### Python DA Assignment 1: NumPy and Pandas

##### Numpy Array Operations: (Python library used for numerical calculations and working with arrays.)
##### Problem Statement: To learn how to use NumPy for creating arrays, performing basic
##### array operations, and conducting numerical computations through analyzing temperature data.
##### Scenario: You are given a list of daily average temperatures (in degrees Celsius)
##### recorded over two weeks. Your task is to use NumPy to analyze this data.

#### 1. Create a 1D NumPy Array:
##### Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values:

```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

#### 2. Inspection and Properties
##### Inspect the shape, data type, and the number of elements in the array.
##### Inspect the shape of the temperatures_w1 array.

```python
print("Shape:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)
```

#### 3. Array Operations:
##### Convert the temperatures_w1 from Celsius to Fahrenheit using the
##### formula: Fahrenheit = (Celsius * 9/5) + 32


```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
```

##### Find the maximum, minimum, and mean temperatures for the week.

```python
print("Maximum Temperature:", np.max(temperatures_w1))

print("Minimum Temperature:", np.min(temperatures_w1))

print("Mean Temperature:", np.mean(temperatures_w1))
```

#### 4. Array Slicing and Indexing:

##### Extract the temperatures for the first three days of the week.

```python
first_three_days = temperatures_w1[0:3]   # It selects index 0, 1 and 2. Index 3 is excluded.
print(temperatures_w1)
print("First Three Days:", first_three_days)
```

##### Extract the temperatures for the weekend (last two days).

```python
weekend_temperatures = temperatures_w1[-2:]  # -2 selects the second-last element, : selects all remaining elements until the end.

print("Weekend Temperatures:", weekend_temperatures)
```

##### Extract the temperatures for the middle three days of the week.

```python
middle_three_days = temperatures_w1[2:5]  # Selects the middle three temperatures (index positions 2, 3, and 4).

print("Middle Three Days:", middle_three_days)
```

#### Step 5: Create a 2D NumPy Array

##### Create a 2D NumPy array named temperatures where each row represents
##### temperatures for a different week:
#####  Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
#####  Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

#### 6. Inspect and Slice the 2D Array:

##### Inspect and print the shape, data type, and total number of elements in the
##### temperatures array.

```python
print("Shape:", temperatures.shape)

print("Data Type:", temperatures.dtype)

print("Total Number of Elements:", temperatures.size)
```

##### Extract the temperatures for each week and the weekends (last two days) for both weeks.

```python
week1 = temperatures[0, :] # 0 selects the first row (Week 1), : selects all columns (all seven days).

week2 = temperatures[1, :]

print("Week 1 Temperatures:", week1)
print("Week 2 Temperatures:", week2)

weekend_week1 = temperatures[0, -2:] # 0 selects Week 1, and -2: selects the last two columns (Saturday and Sunday).
weekend_week2 = temperatures[1, -2:] # 1 selects Week 2, and -2: selects the last two columns (Saturday and Sunday).

print("Week 1 Weekend Temperatures:", weekend_week1)
print("Week 2 Weekend Temperatures:", weekend_week2)
```

### Pandas Series:

### 1. Creating Pandas Series:
##### (Python library used for data analysis and data manipulation.)

##### Create a Pandas Series named marks with the following data:
##### Marks: 95, 92, 89, 85, 80
##### Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd

# Create a Pandas Series with student marks and custom index labels.
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

### 2. Indexing and Slicing:

##### Use the integer index position to access the mark of the 1st rank student.

```python
first_rank_mark = marks.iloc[0] # .iloc[] → Access data using integer positions.
# 0 represents the first position, so it retrieves the mark of the 1st rank student.

print("Mark of 1st Rank Student:", first_rank_mark)
```

##### Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

```python
# Use loc to retrieve the marks of Rank1, Rank2, and Rank3 using their index labels.
top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print("Marks of Top 3 Ranks:")
print(top_three)
```

##### Use the iloc accessor to retrieve the mark of the 3rd rank student.

```python
# 2 represents the third position, so it retrieves the mark of the 3rd rank student.
third_rank_mark = marks.iloc[2]

print("Mark of 3rd Rank Student:", third_rank_mark)
```

##### Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

```python
# Filter the Series to retrieve ranks where the marks are greater than 90.
high_marks = marks[marks > 90]

print("Ranks with Marks Greater Than 90:")
print(high_marks)
```

### 3. Manipulating Series:

#### Modify the mark of the student with the 1st rank to 100.

```python
# Use loc to update the mark of Rank1 to 100.
marks.loc['Rank1'] = 100

print("Updated Marks:")
print(marks)
```

#### Remove the entry corresponding to the last rank from the Series.

```python
# Remove the entry corresponding to Rank5 from the Series.
marks = marks.drop('Rank5')

print("Updated Marks After Removing Rank5:")
print(marks)
```

### Traceback occurred because Rank5 was not found in the Series, as it had already been removed.

```python
print(marks)
```

#### Compute the CGPA by dividing each mark in the Series by 10.

```python
# Divide each student's mark by 10 to calculate the CGPA.
cgpa = marks / 10

print("CGPA of Students:")
print(cgpa)
```

### Pandas DataFrame:

#### 1. Creating Pandas DataFrame:


##### Create a Pandas DataFrame named transactions with the following data:
##### TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
##### ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
##### 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
##### Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'
##### Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
# Creating a dictionary
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Creating DataFrame
transactions = pd.DataFrame(data)

# Displaying DataFrame
transactions
```

#### 2. Data Exploration:

##### Display the transactions DataFrame and its basic information, including
##### head, tail, shape, column names, and data types.

```python

display(transactions) # Display the complete DataFrame

print("First 5 rows:") 
display(transactions.head()) # Display the first 5 rows

print("Last 5 rows:") 
display(transactions.tail()) # Display the last 5 rows


print("Shape of DataFrame:")
print(transactions.shape) # Display the shape (rows and columns)


print("Column Names:")
print(transactions.columns) # Display column names

print("Data Types:")
print(transactions.dtypes) # Display data types
```

##### Display only the 'ProductCategory' and 'Amount' columns.

```python
transactions[['ProductCategory', 'Amount']]
```

##### Retrieve the last 3 columns of the DataFrame using iloc or loc.

```python
# iloc is used for integer-position-based selection
transactions.iloc[:, -3:]  # : selects all row, -3: selects the last 3 columns of the DataFrame.
```

##### Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

```python
transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
```

##### Find the value counts for the 'ProductCategory' column.

```python
transactions['ProductCategory'].value_counts()
```

##### Find the unique values in the 'Region' column.

```python
transactions['Region'].unique()
```

##### Group by 'Region' and find the mean amount for each region.

```python
transactions.groupby('Region')['Amount'].mean()
```

#### 3. Manipulating the DataFrame:

##### Modify the 'Amount' for the transaction with TransactionID 102 to 165.

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

##### Add a new column 'Discount' by calculating 10% of the 'Amount'.

```python
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

##### Remove the row with TransactionID 109.

```python
transactions = transactions[transactions['TransactionID'] != 109]
transactions
```

##### Delete the 'Discount' column from the DataFrame.

```python
transactions.drop('Discount', axis=1, inplace=True)

transactions
```
